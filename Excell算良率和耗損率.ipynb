```python
# 清洗明細資料：跳過前兩列表頭，並移除空白列
df_clean = df_mingxi.iloc[2:].copy()
df_clean.columns = df_mingxi.iloc[1]  # 正確設置欄位名稱
df_clean = df_clean.dropna(subset=["產品編號", "製令數量", "實際完工數"], how='any')  # 去掉非資料列

# 將數值欄位轉換為數字類型
df_clean["製令數量"] = pd.to_numeric(df_clean["製令數量"], errors='coerce')
df_clean["實際完工數"] = pd.to_numeric(df_clean["實際完工數"], errors='coerce')
df_clean["已領用數(基)"] = pd.to_numeric(df_clean["已領用數(基)"], errors='coerce')
df_clean["已耗用數(基)"] = pd.to_numeric(df_clean["已耗用數(基)"], errors='coerce')

# 移除空值列（有些原料欄位會在下一列呈現，不是產品主列）
df_products = df_clean.dropna(subset=["製令數量", "實際完工數"])

# 整體良率計算
total_order_qty = df_products["製令數量"].sum()
total_actual_qty = df_products["實際完工數"].sum()
overall_yield = total_actual_qty / total_order_qty if total_order_qty else None

# 原物料耗損率計算（從所有原料紀錄合計）
total_issued = df_clean["已領用數(基)"].sum()
total_consumed = df_clean["已耗用數(基)"].sum()
waste_rate = 1 - (total_consumed / total_issued) if total_issued else None

overall_yield, waste_rate
```
