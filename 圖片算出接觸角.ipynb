```python
# 使用 OpenCV + SciPy 自動計算接觸角（需安裝庫：pip install opencv-python scipy）
import cv2
import numpy as np
from scipy.optimize import curve_fit

# 1. 讀取圖片並提取輪廓
img = cv2.imread('droplet.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 2. 擬合橢圓並計算接觸角
ellipse = cv2.fitEllipse(contours[0])
(xc, yc), (d1, d2), angle = ellipse  # 橢圓參數
contact_angle = np.degrees(np.arctan2(d2, d1)) * 2  # 計算公式

print(f"接觸角: {contact_angle:.2f}°")
```
