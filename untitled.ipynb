```python

import pandas as pd

def 計算產品良率(df):
    """
    計算每個產品的良率
    :param df: 包含生產數據的DataFrame
    :return: 良率分析結果DataFrame
    """
    # 計算良率
    df['良率(%)'] = (df['實際完工數'] / df['製令數量']) * 100

    # 只保留需要的欄位
    良率分析 = df[['單據編號', '產品編號', '產品批號', '製令數量', '實際完工數', '良率(%)']]

    return 良率分析.sort_values('良率(%)', ascending=False)

def 計算原料耗損率(df):
    """
    計算每種原材料的耗損率
    :param df: 包含原料使用數據的DataFrame
    :return: 耗損率分析結果DataFrame
    """
    # 計算耗損量和耗損率
    df['耗損量'] = df['已領用數(基)'] - df['已耗用數(基)']
    df['耗損率(%)'] = (df['耗損量'] / df['已領用數(基)']) * 100

    # 只保留需要的欄位
    耗損率分析 = df[['原料編號', '原料批號', '已領用數(基)', '已耗用數(基)', '耗損量', '耗損率(%)']]

    return 耗損率分析.sort_values('耗損率(%)', ascending=False)

def 生成分析報告(良率結果, 耗損率結果):
    """
    生成分析報告
    :param 良率結果: 良率分析結果DataFrame
    :param 耗損率結果: 耗損率分析結果DataFrame
    :return: 分析報告HTML
    """
    # 良率分析摘要
    良率摘要 = 良率結果.describe()
    高耗損原料 = 耗損率結果[耗損率結果['耗損率(%)'] > 5]  # 耗損率超過5%的原料

    # 生成HTML報告
    報告 = f"""
    <h2>生產分析報告</h2>

    <h3>1. 產品良率分析</h3>
    {良率結果.to_html()}

    <h4>良率統計摘要</h4>
    {良率摘要.to_html()}

    <h3>2. 原材料耗損率分析</h3>
    {耗損率結果.to_html()}

    <h4>高耗損原料(耗損率>5%)</h4>
    {高耗損原料.to_html()}

    <h3>3. 建議改進措施</h3>
    <ul>
        <li>對高耗損率原材料進行根本原因分析</li>
        <li>對良率低於平均的產品優化生產工藝</li>
        <li>建立耗損率預警機制</li>
        <li>定期統計分析這些指標，持續改進</li>
    </ul>
    """

    return 報告

# 主程式
if __name__ == "__main__":
    # 讀取Excel數據 (假設數據已經整理成適當格式)
    try:
        # 這裡替換成您的實際文件路徑
        df = pd.read_excel('生產數據.xlsx')

        # 計算良率和耗損率
        良率結果 = 計算產品良率(df)
        耗損率結果 = 計算原料耗損率(df)

        # 生成報告
        分析報告 = 生成分析報告(良率結果, 耗損率結果)

        # 保存報告到HTML文件
        with open('生產分析報告.html', 'w', encoding='utf-8') as f:
            f.write(分析報告)

        print("分析完成，報告已生成為 '生產分析報告.html'")

        # 在控制台顯示部分結果
        print("\n良率最高的5個產品:")
        print(良率結果.head())

        print("\n耗損率最高的5種原料:")
        print(耗損率結果.head())

    except Exception as e:
        print(f"發生錯誤: {str(e)}")
```
