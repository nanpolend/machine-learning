```python
import pandas as pd

def calculate_yield_loss(excel_file):
    try:
        # 讀取Excel檔案
        df = pd.read_excel(excel_file)

        # 假設Excel欄位包含：產品名稱、批號、總數量、良品數量
        # 檢查必要的欄位是否存在
        required_columns = ['產品名稱', '批號', '總數量', '良品數量']
        if not all(col in df.columns for col in required_columns):
            return "Excel檔案缺少必要欄位：產品名稱、批號、總數量、良品數量"

        # 按產品名稱分組並計算總和
        grouped = df.groupby('產品名稱').agg({
            '總數量': 'sum',
            '良品數量': 'sum'
        }).reset_index()

        # 計算良率和耗損率
        grouped['良率'] = (grouped['良品數量'] / grouped['總數量'] * 100).round(2)
        grouped['耗損率'] = (100 - grouped['良率']).round(2)

        # 將結果保存到新的Excel檔案
        output_file = '產品良率耗損率統計.xlsx'
        grouped.to_excel(output_file, index=False)

        # 打印結果
        print("\n各產品良率與耗損率統計：")
        print(grouped)
        print(f"\n結果已保存至 {output_file}")

        return grouped

    except Exception as e:
        return f"處理過程中發生錯誤：{str(e)}"

# 使用範例
if __name__ == "__main__":
    # 假設Excel檔案路徑
    file_path = "產品數據.xlsx"  # 替換為實際檔案路徑
    result = calculate_yield_loss(file_path)
```
